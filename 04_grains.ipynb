```python
from semantic_workflows.pyiron.build import bulk, polycrystal
from pyiron_workflow import Workflow
from ase.visualize import view
from semantic_workflows.pyiron.elastic import mechanical_response_test
from ase.visualize import view
from conceptual_dictionary import ConceptualDict 
```

```python
box_size = (15, 15, 15)  
grain_size = 10.0 # in Angstroms
```

```python
pair_style = "eam/alloy"
pair_coeff = "* * workflows/potentials/Fe_Ack.eam Fe"
```

```python
cd = ConceptualDict()

wf = Workflow('x1')
wf.struct = bulk('Fe', cubic=True, cdict=cd)
wf.poly = polycrystal(wf.struct, box_size, grain_size, cdict=cd)
wf.tensile = mechanical_response_test(wf.poly, pair_style, pair_coeff, cdict=cd,
                                     mode='uniaxial') 
```

```python
wf.draw(rankdir='LR', size=(10, 10))
```

```python
wf.run()
```

```python
cd.to_yaml('uniaxial.yaml')
```
