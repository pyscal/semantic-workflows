```python
from semantic_workflows.pyiron.elastic import calculate_elastic_constants
from semantic_workflows.pyiron.build import bulk
from pyiron_workflow import Workflow
from conceptual_dictionary import ConceptualDict 
```

```python
pair_style = "eam/alloy"
pair_coeff = "* * workflows/potentials/Fe_Ack.eam Fe"
```

```python
cd = ConceptualDict()
```

```python
wf = Workflow('e1')
wf.structure = bulk('Fe', cubic=True, cdict=cd)
wf.elastic = calculate_elastic_constants(wf.structure, pair_style, pair_coeff, cdict=cd)
```

```python
wf.draw(rankdir='LR', size=(10, 10))
```

```python
wf.run()
```

```python
cd.to_yaml('elastic.yaml')
```
